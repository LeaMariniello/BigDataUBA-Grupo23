```python
import pandas as pd
import numpy as np
import statsmodels.api as sm
from statsmodels.tools.tools import add_constant
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import confusion_matrix, roc_curve, roc_auc_score, accuracy_score, mean_squared_error, mean_absolute_error
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler # NUEVO: Para escalar características
from sklearn.pipeline import make_pipeline
import matplotlib.pyplot as plt
import seaborn as sns # Ya estaba, se mantiene para visualización

# ─────────────────────── 1. FUNCIÓN DE MAPE0 ───────────────────────────────
def unificar_estado(valor):
    """
    Normaliza los valores de la columna 'ESTADO'/'CH15_COD' a un formato numérico.
    """
    if pd.isna(valor):
        return np.nan

    # Convertir a minúsculas y quitar espacios si es string para un mapeo robusto
    if isinstance(valor, str):
        valor = valor.strip().lower()
        if "ocupado" in valor and "des" not in valor:
            return 1  # Ocupado
        elif "desocupado" in valor:
            return 2  # Desocupado
        elif "inactivo" in valor:
            return 3  # Inactivo
        elif "menor de 10" in valor:
            return 4  # Menor de 10 años
        elif "entrevista individual no realizada" in valor:
            return 0  # Entrevista no realizada (corresponde a 'norespondieron')
        else:
            return np.nan  # Otro caso no mapeado

    # Si ya es numérico, mapear directamente. Esto es para los datos de 2024 que vienen con números.
    try:
        valor_int = int(valor)
        mapa_num = {1: 1, 2: 2, 3: 3, 4: 4, 0: 0}
        return mapa_num.get(valor_int, np.nan)
    except (ValueError, TypeError):
        return np.nan  # No se pudo convertir a numérico

# ─────────────────────── 2. CARGA Y PREPARACIÓN INICIAL DE DATOS ────────────────────────────
def cargar_y_preparar_datos():
    """
    Carga los datasets, unifica nombres de columnas, aplica mapeo de estado,
    filtra por región y concatena. Separa en 'respondieron' y 'norespondieron'.
    """
    # Cargar los datasets. Asegúrate de que las rutas sean correctas.
    t2004 = pd.read_stata(
        'D:/Mis documentos/Estudio/Lic en economia/2do Año/Big data & machine learning/Tps/TP II/Individual_t104.dta',
        convert_categoricals=True
    )
    t2024 = pd.read_excel(
        'D:/Mis documentos/Estudio/Lic en economia/2do Año/Big data & machine learning/Tps/TP II/usu_individual_T124.xlsx'
    )

    # Convertir nombres de columnas a mayúsculas para unificar
    for df_temp in (t2004, t2024):
        df_temp.columns = df_temp.columns.str.upper()

    # Aplicar mapeo de estado para normalizar 'ESTADO'
    t2004["ESTADO"] = t2004["ESTADO"].apply(unificar_estado)
    t2024["ESTADO"] = t2024["ESTADO"].apply(unificar_estado)

    # Filtrar por región "Gran Buenos Aires"
    # Para 2004, 'REGION' es string. Para 2024, 'REGION' es numérico (1 para GBA).
    t2004 = t2004[t2004["REGION"] == "Gran Buenos Aires"].copy()
    t2024 = t2024[t2024["REGION"] == 1].copy()

    # Identificar columnas comunes para concatenar (se recomienda si los modelos se entrenan con ambos años)
    # Si los modelos se entrenan y prueban por año de forma independiente, esta parte es menos crítica.
    columnas_comunes = list(set(t2004.columns) & set(t2024.columns))
    t2004 = t2004[columnas_comunes].copy()
    t2024 = t2024[columnas_comunes].copy()

    # Añadir año a cada DataFrame antes de concatenar
    t2004["ANO_EPH"] = 2004
    t2024["ANO_EPH"] = 2024

    # Concatenar ambos años en un único DataFrame
    df_full = pd.concat([t2004, t2024], ignore_index=True)

    # Crear la variable dependiente 'desocupado' para la base 'respondieron'
    # Solo consideramos casos donde ESTADO no es 0 (entrevista no realizada)
    respondieron = df_full[df_full["ESTADO"] != 0].copy()
    respondieron["desocupado"] = (respondieron["ESTADO"] == 2).astype(int)
    # Convertir ESTADO a int para consistencia, aunque luego se excluye de X
    respondieron["ESTADO"] = respondieron["ESTADO"].astype(int)
    # Eliminar filas donde 'desocupado' es NaN (si el mapeo generó alguno)
    respondieron = respondieron.dropna(subset=["desocupado"])

    # Base de 'norespondieron' (donde ESTADO es 0)
    norespondieron = df_full[df_full["ESTADO"] == 0].copy()

    return respondieron, norespondieron

# ─────────────────────── 3. TRANSFORMACIONES Y PREPARACIÓN DE MATRICES X/Y ───────────────────────────────
def procesar_datos(df_input):
    """
    Realiza la ingeniería de características y codificación one-hot para los datos.
    Excluye variables que causan fuga de datos o son identificadores.
    """
    df = df_input.copy() # Trabajar con una copia para evitar SettingWithCopyWarning

    # Variables a EXCLUIR de la matriz X.
    # CRÍTICO: 'CH15_COD' es la variable original que corresponde a 'ESTADO'.
    # 'ESTADO' y 'desocupado' son las variables objetivo o sus proxies.
    # Las variables de ID o texto irrelevantes también se excluyen.
    excluir_from_X = [
        'desocupado', 'ANO_EPH', 'CODUSU', 'ANO4', 'TRIMESTRE',
        'NRO_HOGAR', 'COMPONENTE', 'REGION', 'ESTADO', 'CH15_COD', # <-- Fuga de datos si se incluye!
        'CH15_NOM', 'CH16_COD', 'CH16_NOM', 'CH17_COD', 'CH17_NOM',
        'PP02C8', 'PP02C4', 'PP02C7', 'PP02C1', 'PP02C5',
        'PP02C3', 'PP02C6', 'PP02C2', 'PP11B1'
    ]

    # Eliminar las columnas especificadas de `excluir_from_X` del DataFrame.
    # `errors='ignore'` evita errores si una columna no existe.
    df = df.drop(columns=[col for col in excluir_from_X if col in df.columns], errors='ignore')

    # Identificar todas las columnas categóricas que deben ser one-hot encoded.
    # Esto incluye las que ya estaban en tu lista y las adicionales que pueden ser categóricas.
    # No deben estar en `excluir_from_X`.
    categorical_features_to_encode = [
        'PP08D1', 'PP06D', 'PP06C', 'PP08J2', 'PP03D', 'PP04B2',
        'CAT_INAC', 'PP07K', 'PP07H', 'PP07I', 'PP07J',
        'CH04', 'CH05', 'CH06', 'CH07', 'CH08', 'CH09', 'CH10', 'CH11', 'CH12', 'CH13',
        'CH14', 'CH16', 'CH17', 'CH18', 'CH19', 'CH20', 'CH21', 'CH22', 'CH23', 'CH24',
        'CH25', 'CH26', 'CH27', 'CH28', 'CH29', 'CH30', 'CH31', 'CH32', 'CH33', 'CH34'
    ]

    # Convertir estas columnas a tipo 'category' para que `pd.get_dummies` las procese.
    for col in categorical_features_to_encode:
        if col in df.columns:
            df[col] = df[col].astype('category')
    
    # NUEVO: Realizar la codificación one-hot para todas las columnas de tipo 'category'.
    # `drop_first=True` evita la multicolinealidad perfecta (dummy trap).
    df = pd.get_dummies(df, drop_first=True)

    # Convertir cualquier columna restante de tipo 'object' a numérica.
    # Esto manejaría casos donde alguna columna inesperada sea string y se deba convertir.
    for col in df.columns:
        if df[col].dtype == 'object':
            df[col] = pd.to_numeric(df[col], errors='coerce')

    # Eliminar columnas que estén completamente vacías (todo NaN).
    df = df.dropna(axis=1, how='all')

    # Separar la matriz X (características) y el vector y (variable objetivo).
    # 'desocupado' solo estará presente en el DataFrame 'respondieron'.
    if 'desocupado' in df_input.columns: # Usamos df_input para verificar si el original tenía 'desocupado'
        y = df_input['desocupado'].astype(int) # 'y' se extrae del df_input original para asegurar consistencia
                                                # y_full_year se filtrará en split_y_diferencias si tiene NaN
        # X debe contener solo columnas numéricas y no el target
        X_final_cols = [col for col in df.columns if pd.api.types.is_numeric_dtype(df[col]) and col not in ['desocupado']]
        X = df[X_final_cols].copy()
        return X, y
    else:
        # Para el DataFrame 'norespondieron', que no tiene 'desocupado'
        X_final_cols = [col for col in df.columns if pd.api.types.is_numeric_dtype(df[col])]
        X = df[X_final_cols].copy()
        return X

def split_y_diferencias(df_base, año):
    """
    Divide el DataFrame de un año específico en conjuntos de entrenamiento y prueba,
    y calcula las diferencias de medias entre ellos.
    """
    df_año = df_base[df_base['ANO_EPH'] == año].copy()

    # Procesar datos para obtener la matriz X y el vector y.
    # Aquí es donde se aplican todas las transformaciones y la codificación one-hot.
    X_full_year, y_full_year = procesar_datos(df_año)

    # Filtrar columnas con varianza cero en el conjunto de X del año completo.
    # Esto asegura que todos los splits (train/test) tengan las mismas columnas
    # y que no haya columnas problemáticas con varianza 0.
    variances = X_full_year.var()
    cols_to_keep = variances[variances > 0].index
    X_full_year_filtered = X_full_year[cols_to_keep].copy()

    # Dividir en conjuntos de entrenamiento y prueba.
    # 'stratify=y_full_year' es crucial para mantener el balance de clases en ambos conjuntos.
    X_train, X_test, y_train, y_test = train_test_split(
        X_full_year_filtered, y_full_year, test_size=0.3, random_state=444, stratify=y_full_year
    )

    # Calcular y comparar las medias de las variables entre los conjuntos de entrenamiento y prueba.
    diff = pd.DataFrame({
        'Media Entrenamiento': X_train.mean(),
        'Media Prueba': X_test.mean(),
        'Diferencia': X_train.mean() - X_test.mean()
    }).sort_values('Diferencia', key=abs, ascending=False)

    return X_train, X_test, y_train, y_test, diff

# ─────────────────────── 4. EJECUCIÓN CONSIGNAS 1 ──────────────────────────────────────
print("Cargando y preparando datos iniciales...")
respondieron, norespondieron = cargar_y_preparar_datos()

print("\n--- Consigna 1: Diferencias de Medias entre Entrenamiento y Prueba ---")
# Obtener los conjuntos de datos para 2004 y calcular diferencias
X_train_2004, X_test_2004, y_train_2004, y_test_2004, diff_2004 = split_y_diferencias(respondieron, 2004)
print("\nDiferencias 2004 (Top 20 por valor absoluto de la diferencia):")
print(diff_2004.head(20))

# Obtener los conjuntos de datos para 2024 y calcular diferencias
X_train_2024, X_test_2024, y_train_2024, y_test_2024, diff_2024 = split_y_diferencias(respondieron, 2024)
print("\nDiferencias 2024 (Top 20 por valor absoluto de la diferencia):")
print(diff_2024.head(20))

# Verificar el balance de clases en la variable objetivo 'desocupado' para cada año
print("\nBalance de clases en 'desocupado' por año (base 'respondieron'):")
print(respondieron.groupby("ANO_EPH")["desocupado"].value_counts())
```

```python
import pandas as pd
import numpy as np
import statsmodels.api as sm
from statsmodels.tools.tools import add_constant

def preparar_datos_para_modelos(df):
    df = df.copy()
    
    df['edad'] = pd.to_numeric(df['CH06'], errors='coerce')
    df['edad2'] = df['edad']**2
    df['mujer'] = (df['CH04'] == 2).astype(int)
    
    def calcular_educ(row):
        valor = row['CH12']
        
        niveles_texto = {
            'jardín/preescolar': 0,
            'educación especial (discapacitado)': 0,
            'primario': 0,
            'egb': 0,
            'secundario': 7,
            'polimodal': 7,
            'terciario': 13,
            'universitario': 13,
            'posgrado universitario': 18,
        }
        
        niveles_num = {
            0: 0, 1:0, 2:3, 3:7, 4:10, 5:12, 6:14, 7:16, 8:16, 9:17, 10:19,
            99: np.nan
        }
        
        try:
            valor_num = float(valor)
            if valor_num in niveles_num:
                return niveles_num[valor_num]
        except:
            pass
        
        if isinstance(valor, str):
            v_lower = valor.strip().lower()
            if v_lower in niveles_texto:
                return niveles_texto[v_lower]
        
        return np.nan
    
    df['educ'] = df.apply(calcular_educ, axis=1)
    df['tiene_cobertura'] = (df['CH08'] == 1).astype(int)
    
    # Agregamos la nueva variable estado_civil desde CH07
    df['estado_civil'] = pd.to_numeric(df['CH07'], errors='coerce')
    
    ipc_base = 100
    ipc_actual = 235.125
    factor_ajuste = ipc_actual / ipc_base
    df['salario_semanal'] = (pd.to_numeric(df['P21'], errors='coerce') * factor_ajuste) / (52/12)
    
    q_low = df['salario_semanal'].quantile(0.01)
    q_hi = df['salario_semanal'].quantile(0.99)
    df = df[(df['salario_semanal'] > 0) & 
            (df['salario_semanal'] >= q_low) & 
            (df['salario_semanal'] <= q_hi)].copy()
    
    df['log_salario'] = np.log(df['salario_semanal'])
    
    return df


def estimar_modelos(train_data):
    """Estima los modelos solicitados en la consigna 2 y devuelve los modelos entrenados"""
    # Configuración de modelos según consigna
    modelos_config = [
    ['edad'],
    ['edad', 'edad2'],
    ['edad', 'edad2', 'educ'],
    ['edad', 'edad2', 'educ', 'mujer'],
    ['edad', 'edad2', 'educ', 'mujer', 'tiene_cobertura', 'estado_civil']
    ]

    resultados = []
    modelos_entrenados = []

    for i, vars_modelo in enumerate(modelos_config, 1):
        try:
            # Seleccionar datos y eliminar NA
            datos = train_data[vars_modelo + ['log_salario']].dropna()
            
            if len(datos) < 10:
                print(f"Modelo {i} omitido - solo {len(datos)} observaciones válidas")
                modelos_entrenados.append(None)
                continue
                
            # Ajustar modelo
            X = add_constant(datos[vars_modelo])
            y = datos['log_salario']
            modelo = sm.OLS(y, X).fit()
            modelos_entrenados.append(modelo)
            
            # Preparar resultados para tabla
            fila = {'Modelo': f"({i})", 'N': len(datos), 'R2': round(modelo.rsquared, 3)}
            
            # Agregar coeficientes con formato requerido
            for var in ['const'] + vars_modelo:
                coef = modelo.params[var]
                se = modelo.bse[var]
                p = modelo.pvalues[var]
                stars = '***' if p < 0.001 else '**' if p < 0.05 else '*' if p < 0.1 else ''
                fila[var] = f"{coef:.3f}{stars} ({se:.2f})"
            
            # Rellenar con NA las variables no usadas en este modelo
            todas_vars = ['edad', 'edad2', 'educ', 'mujer', 'tiene_cobertura', 'estado_civil']
            for var in todas_vars:
                if var not in fila:
                    fila[var] = "NA"
            
            resultados.append(fila)
            
        except Exception as e:
            print(f"Error en modelo {i}: {str(e)}")
            modelos_entrenados.append(None)
            continue
    
    return resultados, modelos_entrenados

def generar_tabla_resultados(resultados):
    """Genera la tabla de resultados según formato solicitado"""
    columnas = ['Modelo', 'edad', 'edad2', 'educ', 'mujer',
               'tiene_cobertura', 'estado_civil', 'N', 'R2']
    
    tabla_final = pd.DataFrame(resultados, columns=columnas)
    
    print("\n" + "="*80)
    print("Tabla 2. Estimación por regresión lineal de salarios usando la base de entrenamiento")
    print("Var. Dep: log(salario_semanal)")
    print("Nota: * p<0.1, ** p<0.05, *** p<0.001")
    print("="*80)
    
    return tabla_final

# -----------------------------------------------------------
# EJECUCIÓN PRINCIPAL
# -----------------------------------------------------------

# 1. Cargar datos (asumiendo que 'respondieron' o 'df_gba' está disponible)
if 'respondieron' in globals():
    df_principal = respondieron
elif 'df_gba' in globals():
    df_principal = df_gba
else:
    raise ValueError("No se encontró ni 'respondieron' ni 'df_gba' en los datos")

# 2. Filtrar solo ocupados (ESTADO == 1)
ocupados = df_principal[df_principal['ESTADO'] == 1].copy()

# 3. Dividir en train/test (70/30)
train_data = ocupados.sample(frac=0.7, random_state=444)
test_data = ocupados.drop(train_data.index)

# 4. Procesar datos
print("Preparando datos...")
ocupados_train = preparar_datos_para_modelos(train_data)
ocupados_test = preparar_datos_para_modelos(test_data)

# 5. Estimar modelos y generar tabla
print("Estimando modelos...")
resultados, modelos_entrenados = estimar_modelos(ocupados_train)
tabla_final = generar_tabla_resultados(resultados)

# Mostrar tabla
print(tabla_final.to_string(index=False, float_format=lambda x: f"{x:.3f}"))

# Estadísticas descriptivas
print("\nEstadísticas descriptivas:")
print(f"- Muestra final: {len(ocupados_train)} ocupados en entrenamiento")
print(f"- Salario semanal promedio: ${ocupados_train['salario_semanal'].mean():,.2f}")
print(f"- Edad promedio: {ocupados_train['edad'].mean():.1f} años")
print(f"- Porcentaje de mujeres: {ocupados_train['mujer'].mean()*100:.1f}%")
print(f"- Porcentaje con cobertura médica: {ocupados_train['tiene_cobertura'].mean()*100:.1f}%")
print(f"- Porcentaje de casados: {ocupados_train['estado_civil'].mean()*100:.1f}%")
```

```python
# ─────────────────────── CONSIGNAS 5 Y 6: CLASIFICACIÓN ────────────────────────────

print("\n--- Consigna 5: Métodos de Clasificación y Performance ---")

# Función para evaluar modelos de clasificación con métricas completas y ROC.
# Incluye la Matriz de Confusión, AUC y Accuracy.
def evaluar_modelo_clasificacion(X_train, y_train, X_test, y_test, modelo_pipeline, nombre_modelo):
    """
    Entrena un modelo de clasificación (pipeline) y evalúa su rendimiento
    en el conjunto de prueba, reportando matriz de confusión, AUC y Accuracy.
    También genera un punto en el gráfico ROC.
    """
    # Entrenar el pipeline completo (imputer, scaler, y clasificador)
    modelo_pipeline.fit(X_train, y_train)

    # Realizar predicciones:
    # y_prob: Probabilidad de que la clase sea 1 (desocupado) para cada observación.
    # y_pred: Clasificación binaria (0 o 1) usando un umbral de probabilidad > 0.5.
    y_prob = modelo_pipeline.predict_proba(X_test)[:, 1]
    y_pred = (y_prob > 0.5).astype(int)

    # Calcular métricas de rendimiento
    cm = confusion_matrix(y_test, y_pred) # Matriz de Confusión
    fpr, tpr, _ = roc_curve(y_test, y_prob) # Curva ROC (False Positive Rate, True Positive Rate)
    auc_score = roc_auc_score(y_test, y_prob) # Área bajo la curva ROC (AUC)
    accuracy = accuracy_score(y_test, y_pred) # Accuracy (proporción de predicciones correctas)

    # Imprimir los resultados en la consola
    print(f"\nModelo: {nombre_modelo}")
    print("Matriz de Confusión:\n", cm)
    print(f"AUC: {auc_score:.4f}")
    print(f"Accuracy: {accuracy:.4f}")

    # Plotear la curva ROC en el gráfico compartido
    plt.plot(fpr, tpr, label=f'{nombre_modelo} (AUC = {auc_score:.3f})')

    # Devolver los resultados en un diccionario
    return {
        'modelo': nombre_modelo,
        'auc': auc_score,
        'accuracy': accuracy,
        'matriz_confusion': cm,
        'fpr': fpr, 'tpr': tpr # Se devuelven para que se puedan dibujar en un gráfico centralizado
    }

# Definir los pipelines para los modelos de clasificación.
# Cada pipeline incluye:
# 1. SimpleImputer: Rellena los valores NaN (usando la mediana de la columna).
# 2. StandardScaler: Escala las características para que tengan media 0 y desviación estándar 1.
#    Esto es CRÍTICO para KNN y Logistic Regression, ya que son sensibles a la escala.
# 3. El clasificador (LogisticRegression o KNeighborsClassifier).
#    max_iter en LogisticRegression se aumenta para asegurar la convergencia en datasets grandes.
logit_pipeline_2004 = make_pipeline(SimpleImputer(strategy='median'), StandardScaler(), LogisticRegression(random_state=444, max_iter=5000))
knn_pipeline_2004 = make_pipeline(SimpleImputer(strategy='median'), StandardScaler(), KNeighborsClassifier(n_neighbors=5))

logit_pipeline_2024 = make_pipeline(SimpleImputer(strategy='median'), StandardScaler(), LogisticRegression(random_state=444, max_iter=5000))
knn_pipeline_2024 = make_pipeline(SimpleImputer(strategy='median'), StandardScaler(), KNeighborsClassifier(n_neighbors=5))

# Configurar el gráfico para las Curvas ROC
plt.figure(figsize=(10, 8))
plt.plot([0, 1], [0, 1], 'k--', label='Clasificador Aleatorio (AUC = 0.5)') # Línea de base para un clasificador aleatorio
plt.title("Curvas ROC - Modelos de Clasificación por Año")
plt.xlabel("False Positive Rate (FPR)")
plt.ylabel("True Positive Rate (TPR)")
plt.grid(True, alpha=0.3) # Añadir una grilla para mejor lectura

# Listas para almacenar los resultados de la evaluación de cada modelo
resultados_clasificacion_2004 = []
resultados_clasificacion_2024 = []

# Evaluar modelos para el año 2004
resultados_clasificacion_2004.append(evaluar_modelo_clasificacion(X_train_2004, y_train_2004, X_test_2004, y_test_2004, logit_pipeline_2004, "Logit 2004"))
resultados_clasificacion_2004.append(evaluar_modelo_clasificacion(X_train_2004, y_train_2004, X_test_2004, y_test_2004, knn_pipeline_2004, "KNN 2004"))

# Evaluar modelos para el año 2024
resultados_clasificacion_2024.append(evaluar_modelo_clasificacion(X_train_2024, y_train_2024, X_test_2024, y_test_2024, logit_pipeline_2024, "Logit 2024"))
resultados_clasificacion_2024.append(evaluar_modelo_clasificacion(X_train_2024, y_train_2024, X_test_2024, y_test_2024, knn_pipeline_2024, "KNN 2024"))

# Finalizar y mostrar el gráfico ROC
plt.legend(loc='lower right', fontsize=10) # Colocar la leyenda en la esquina inferior derecha
plt.tight_layout() # Ajustar para evitar que los elementos se superpongan
plt.show()
```
